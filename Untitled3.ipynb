```python
%%writefile /content/proj/src/protocol.py
"""Project protocol: MIT-BIH only, patient-wise split 70/15/15 decided from raw annotations (before any training)."""
import os, json, numpy as np, pandas as pd, wfdb

MITDB = '/content/drive/MyDrive/ecglab/data/mitdb'
SPLIT_FILE = '/content/drive/MyDrive/ecglab/data/split_v2_70_15_15.json'
PACED = [102, 104, 107, 217]                     # paced records are excluded
SAME_PERSON = {202: 201}                          # records 201 and 202 are the same person
CLASSES = ['N', 'S', 'V', 'F', 'Q']
AAMI = {**dict.fromkeys(['N', 'L', 'R', 'e', 'j'], 'N'), **dict.fromkeys(['A', 'a', 'J', 'S'], 'S'),
        **dict.fromkeys(['V', 'E'], 'V'), 'F': 'F', **dict.fromkeys(['/', 'f', 'Q'], 'Q')}
SHARE = {'train': .70, 'val': .15, 'test': .15}

def all_records():
    recs = sorted(int(f[:-4]) for f in os.listdir(MITDB) if f.endswith('.hea') and f[:-4].isdigit())
    return [r for r in recs if r not in PACED]

def beat_counts():
    rows = {}
    for r in all_records():
        a = wfdb.rdann(f'{MITDB}/{r}', 'atr'); c = np.zeros(5, int)
        for s in a.symbol:
            if s in AAMI: c[CLASSES.index(AAMI[s])] += 1
        rows[r] = c
    df = pd.DataFrame(rows).T; df.columns = CLASSES
    return df

def make_split():
    """Deterministic greedy allocation of persons (not beats) to train/val/test, matching class shares."""
    cnt = beat_counts(); cnt['person'] = [SAME_PERSON.get(r, r) for r in cnt.index]
    g = cnt.groupby('person')[['N', 'S', 'V', 'F']].sum(); tot = g.sum()
    names = list(SHARE); share = np.array(list(SHARE.values()))
    alloc, n, assign = np.zeros((3, 4)), np.zeros(3), {}
    order = (g[['S', 'V', 'F']] / tot[['S', 'V', 'F']]).sum(1).sort_values(ascending=False).index
    for k in order:                               # persons rich in rare classes first
        costs = []
        for i in range(3):
            a2, n2 = alloc.copy(), n.copy(); a2[i] += g.loc[k].values; n2[i] += 1
            costs.append((((a2 - share[:, None] * tot.values) / tot.values) ** 2).sum()
                         + (((n2 - share * len(g)) / len(g)) ** 2).sum())
        i = int(np.argmin(costs)); assign[int(k)] = names[i]; alloc[i] += g.loc[k].values; n[i] += 1
    return assign, g, alloc, n

def record_split(assign):
    return {int(r): assign[SAME_PERSON.get(r, r)] for r in all_records()}
```

```python
%%writefile /content/proj/src/data.py
"""Raw MIT-BIH -> filtered beats + RR features + record id + split label (from splits json)."""
import json, numpy as np, wfdb
from scipy.signal import butter, sosfiltfilt
import protocol
from protocol import AAMI, CLASSES, MITDB

FS, PRE, POST = 360, 90, 110            # beat window: 90 samples before the R-peak, 110 after

def bandpass(x, fs=FS, lo=0.5, hi=40, order=3):
    sos = butter(order, [lo, hi], btype='bandpass', fs=fs, output='sos')
    return sosfiltfilt(sos, x)           # zero-phase

def preprocess(x, fs=FS):
    x = bandpass(x, fs)
    return (x - x.mean()) / (x.std() + 1e-8)     # z-score per record

def lead_index(r):
    names = [n.upper() for n in r.sig_name]
    return names.index('MLII') if 'MLII' in names else 0

def process_record(rec):
    r = wfdb.rdrecord(f'{MITDB}/{rec}'); a = wfdb.rdann(f'{MITDB}/{rec}', 'atr')
    sig = preprocess(r.p_signal[:, lead_index(r)], r.fs)
    keep = [(s, AAMI[sym]) for s, sym in zip(a.sample, a.symbol) if sym in AAMI]
    pos = np.array([s for s, _ in keep]); lab = [l for _, l in keep]
    rr_all = np.diff(pos); B, RR, Y = [], [], []
    for i in range(1, len(pos) - 1):
        if pos[i] - PRE < 0 or pos[i] + POST > len(sig):
            continue
        pre_rr, post_rr = rr_all[i - 1], rr_all[i]
        local = np.mean(rr_all[max(0, i - 10):i])                 # previous intervals only
        hist = rr_all[max(0, i - 21):i - 1]
        med = np.median(hist) if len(hist) else pre_rr
        RR.append([pre_rr / r.fs, post_rr / r.fs, pre_rr / post_rr, local / r.fs,
                   pre_rr / med, post_rr / med, local / med])
        B.append(sig[pos[i] - PRE:pos[i] + POST]); Y.append(CLASSES.index(lab[i]))
    return np.array(B, dtype='float32'), np.array(RR, dtype='float32'), np.array(Y)

def build_all(out_npz):
    split = {int(k): v for k, v in json.load(open(protocol.SPLIT_FILE)).items()}
    X, RR, Y, REC, SPL = [], [], [], [], []
    for rec in protocol.all_records():
        B, R, y = process_record(rec)
        X.append(B); RR.append(R); Y.append(y)
        REC.append(np.full(len(y), rec)); SPL.append(np.full(len(y), split[rec]))
    out = dict(X=np.concatenate(X), RR=np.concatenate(RR), y=np.concatenate(Y),
               rec=np.concatenate(REC), split=np.concatenate(SPL))
    np.savez_compressed(out_npz, **out)
    return out
```

```python
%%writefile /content/proj/src/features.py
"""Beat features: 12 baseline (4 RR + 8 beat-shape) and 15 (adds 3 RR features normalized by the patient's own rhythm)."""
import numpy as np
from scipy.stats import skew, kurtosis

FS, R0 = 360, 90          # R-peak index inside the 200-sample window

def qrs_width(b):
    base = np.median(b); half = 0.5 * abs(b[R0] - base); dev = np.abs(b - base) >= half
    l = r = R0
    while l > 0 and dev[l - 1]: l -= 1
    while r < len(b) - 1 and dev[r + 1]: r += 1
    return (r - l + 1) / FS

def beat_features(b):
    base = np.mean(b[R0 - 30:R0 - 15]); st = np.mean(b[R0 + 30:R0 + 45]) - base
    return [qrs_width(b), b[R0], st, b.std(), skew(b), kurtosis(b), b.max(), b.min()]

NAMES12 = ['pre_rr', 'post_rr', 'rr_ratio', 'local_rr', 'qrs_width', 'r_amp', 'st_dev', 'std', 'skew', 'kurt', 'max', 'min']
NAMES15 = NAMES12 + ['pre_rr_norm', 'post_rr_norm', 'local_rr_norm']

def build(X, RR):
    M = np.array([beat_features(b) for b in X], dtype='float32')
    F12 = np.hstack([RR[:, :4], M]); F15 = np.hstack([F12, RR[:, 4:7]])
    return F12, F15
```

```python
import numpy as np, pandas as pd
from imblearn.pipeline import Pipeline
from imblearn.under_sampling import RandomUnderSampler
from imblearn.over_sampling import SMOTE
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.utils.class_weight import compute_sample_weight
from sklearn.metrics import precision_recall_fscore_support as prfs, accuracy_score
from xgboost import XGBClassifier

D = '/content/drive/MyDrive/ecglab/data'
d = np.load(f'{D}/v2_features.npz', allow_pickle=True)
y, split = d['y'], d['split']
tr, va = (split == 'train') & (y < 4), (split == 'val') & (y < 4)       # class Q excluded
print('val beats per class [N,S,V,F]:', np.bincount(y[va], minlength=4))

def bal():
    return [('under', RandomUnderSampler(sampling_strategy={0: 10000}, random_state=42)),
            ('smote', SMOTE(sampling_strategy={1: 4000, 3: 2000}, k_neighbors=5, random_state=42))]
def make(name):
    if name == 'SVM':
        return Pipeline([('sc', StandardScaler()), *bal(), ('clf', SVC(kernel='rbf', C=10, gamma='scale', random_state=42))])
    if name == 'RF':
        return Pipeline([('sc', StandardScaler()), *bal(), ('clf', RandomForestClassifier(n_estimators=300, n_jobs=-1, random_state=42))])
    return XGBClassifier(n_estimators=300, max_depth=6, learning_rate=0.1, subsample=0.8, colsample_bytree=0.8,
                         tree_method='hist', n_jobs=-1, random_state=42)

rows = []
for fs_name, F in [('12 feat', d['F12']), ('15 feat', d['F15'])]:
    for name in ['SVM', 'RF', 'XGBoost']:
        m = make(name)
        if name == 'XGBoost':
            m.fit(F[tr], y[tr], sample_weight=np.clip(compute_sample_weight('balanced', y[tr]), None, 50))
        else:
            m.fit(F[tr], y[tr])
        p = m.predict(F[va])
        _, rc, f, sp = prfs(y[va], p, labels=[0, 1, 2, 3], zero_division=0)
        rows.append(dict(features=fs_name, model=name, acc=accuracy_score(y[va], p),
                         F1_N=f[0], F1_S=f[1], F1_V=f[2], F1_F=f[3], S_rec=rc[1], V_rec=rc[2],
                         meanSV=(f[1] + f[2]) / 2))
res = pd.DataFrame(rows).round(3)
print(res.sort_values('meanSV', ascending=False).to_string(index=False))
res.to_csv('/content/proj/reports/val_stage4.csv', index=False)
```
